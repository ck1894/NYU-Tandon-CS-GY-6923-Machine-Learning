## Creating vectors
We can create vectors a number of ways.  We can manually create a vector by specifying the elements from a list.  Note that, unlike MATLAB, there is no difference between a row and column vector.  Also, you have to call the `print` command to print the object.

```python
x = np.array([1,2,4])
print(x)
```

You can also create a vector containing a range of numbers.

```python
x1 = np.arange(10)   # numbers from 0 to 9 (note 10 is NOT included)
x2 = np.arange(2,7)  # numbers from 2 to 6
print("x1 = "+str(x1))
print("x2 = "+str(x2))
```

You can use a third `step` argument.  The step does not need to be positive or integer.

```python
x1 = np.arange(10,50,5)   # Increments of 5.  Note this ends on 45, not 50
print(x1)
```

```python
# Increments of 2.5.  To ensure it includes the value 10, we make the endpoint slightly >10.
step = 2.5
x2 = np.arange(0,10+.1,step)
print(x2)
```

Most operations can be vectorized meaning that the operation is applied to each component.

```python
x = np.arange(2,6)
y = x**2
z = 2**x
print("x = " + str(x))
print("y = " + str(y))
print("z = " + str(z))
```

## Indexing and Slicing Vectors

You can access individual elements by `x[i]`.  Unlike MATLAB, the indexing starts at 0!  

```python
x = np.array([2,4,6,8,10])
print("Element 0 is "+str(x[0]))
print("Element 3 is "+str(x[3]))
```

If you want to get the last element, it might help to know the array length. This can be done using the "shape" attribute. Note that this returns a tuple with 2 elements by default because numpy also allows us to work with matrices that have both a length and a height.

```python
print(x.shape)
print(x.shape[0])
```

Similar to MATLAB, you can obtain *slices* from `numpy` vectors, meaning subsets of components.  Use the `:` operator as follows.

```python
x = np.random.uniform(0,1,10)  # 10 random elements from 0 to 1
xlength = x.shape[0]
x1 = x[2:5]             # Elements 2,3,4  (Note 5 is NOT included)
x2 = x[:4]              # Elements 0,1,2,3  (Starts at 0, element 4 NOT included)
x3 = x[7:]              # Elements 7,8,9 (Element 7 IS included. Then all elements to the end.)
xlast = x[-1]           # The last element
# or you could do xlast = x[xlength-1]
x4 = x[2:-3]            # Elements 3,4,5,6
# This is equivlent to writing x[2:xlength-3]

print("x=      "+np.array_str(x))
print("length of x =      "+str(xlength))
print("x[2:5]= "+str(x1))
print("x[:4]=  "+str(x2))
print("x[7:]=  "+str(x3))
print("x[-1]=  "+str(xlast))
print("x[2:-3]=  "+str(x4))
```

# Numpy 2D Arrays (aka matrices)

In addition to vectors, we will also need to work with matrix data, which in numpy are represented as 2 dimensional arrays.

We will start with a simple example of how to create a 2-d array (aka a matrix).

```python
X = np.array([[0,1],[2,3],[4,5]])
print(X)
```

You can check `X`'s dimensions using the `shape` variable. Do this often! One of the most common mistakes is having a dimension mismatch.

numpy also handles vectors is a sloppy way. There are both 1-d arrays with no second dimension, and 2-d arrays with one of the dimensions equal to 1. Be aware!

```python
X.shape
```

```python
y = np.array([6,7])
y.shape
```

```python
y = np.array([[6],[7]])
y.shape
```

To make a 1-d into a 2-d array with dimension one, use the `None` operation. You can either create a row or column vector.

```python
y = np.array([6,7])
z = y[None,:]
w = y[:,None]
print(y)
print(y.shape)
print(z)
print(z.shape)
print(w)
print(w.shape)
```

We can perform basic operations on the matrix, like take its transpose, or sum up all elements.

```python
Y = np.transpose(X)
print(Y)
```

```python
np.sum(X)
```

We can also use slice indexing, just like we did for vectors. Note that in the last example we are grabbing the first column from the matrix. This gets returned as a 1-d array, not a 2-d array with second dimenion 1.

```python
print(X);print()
print(X[0,0]);print()
print(X[0,1]);print()
print(X[1,0]);print()
print(X[0:2,0]);print()
print(X[:,0]);print()
```

## Axis Parameter

Many operations in the `numpy` package can take an optional `axis` parameter to specify which dimensions the operation is to be applied.  This is extremely useful for multi-dimensional data. An operation like `np.mean` or `np.sum` takes the mean or sum of *all* elements in the array -- from all rows and columns.

```python
print(np.mean(X))
print(np.sum(X))
```

To take only the `sum` along each column, we can use the `axis` parameter.

```python
print(np.sum(X,axis=0))
```

Since `X` has shape `(3,2)`, the output `np.sum(X,axis=0)` is of shape `(2,)`.  Similarly, we can take the `sum` along each row:

```python
print(np.sum(X,axis=1))
```

## Broadcasting

**Broadcasting** is a useful tool in Python for performing operations on matrices. It generalizes the useful fact that, if we multiply a numpy array by a scalar, Python knows that we want to multiply *every entry* by that scalar.  

```python
a = np.array([[1,2,3],[4,5,6]])
b = 2
print(b*a)
print(a*b)
```

### Example 1:  Mean Removal

Suppose that `X` is a data matrix of shape `(n,d)`.  That is, there are `n` data points and `d` features per point.  Often, we have to remove the mean from each feature.  That is, we want to compute the mean for each feature and then remove the mean from each column.  We could do this with a for-loop as:

```python
# Generate some random data
n = 100
d = 5
X = np.random.rand(n,d)
print(X[0:7,:]) # print the first several rows of X
```

```python
Xm = np.zeros(d)      # Mean for each feature
X_demean = np.zeros((n,d))  # Transformed features with the means removed
for j in range(d):
    Xm[j] = np.mean(X[:,j]) # mean of column j across all rows
    for i in range(n):
        X_demean[i,j] = X[i,j] - Xm[j] # subtract mean from each row in column j
print(X_demean[0:7,:]) # print the first several rows of the result to compare to later
```

The code below does this without a for loop using the `axis` parameter and broadcasting.

```python
# Compute the mean per column using the axis command
Xm = np.mean(X,axis=0)  # This is a d-dim matrix
print(Xm)
```

To use broadcasting we will need to convert Xm to a 2 dimensional ndarray. As show before, we can do this with the `Xm[None,:]` operation, which returns a `(1,d)` shape array.

```python
print(Xm[None,:])
```

```python
Xm.shape
```

```python
Xm[None,:].shape
```

```python
X.shape
```

Using Python broadcasting, we can then subtract the `Xm[None,:]` from `X`. These array are different sizes -- `Xm[None,:]` has one row while `X` has n. But numpy automatically figures out that, since the number of columns match, we want to substract `Xm[None,:]` off of every row in X.

```python
# Subtract the mean
X_demean = X - Xm[None,:]
print(X_demean[0:7,:])
```

### Example 2:  Standardizing variables

A variant of the above example is to *standardize* the features, where we compute the transform variables,

    Z[i,j] = (X[i,j] - Xm[j])/ Xstd[j]
    
where `Xstd[j]` is the standard deviation per feature.  This can be done as follows:

```python
Xstd = np.std(X,axis=0)
Z = (X-Xm[None,:])/Xstd[None,:]
```

## Matrix operations with numpy
Python broadcasting is great, but sometime it does exactly the wrong thing. If you have a column vector `z` and a matrix `X`, `X*z` won't compute the matrix-vector product, but rather use broadcasting to scales `X`'s rows. Or it will throw an error if the dimensions don't work out. Similarly for matrix `X` and `Y`, `X*Y` does not compute a matrix product.

```python
X = np.array([[1,2],[3,4]])
z = np.array([[2],[3]])
Y = np.array([[-1,-1],[-1,-1]])
print(X);print()
print(z);print()
print(Y);print()
```

```python
print(str(X*z));print()
print(X*Y)
```

To compute actually matrix/matrix and matrix/vector products, you can use the `dot` operation.

```python
print(np.dot(X,z));print()
print(np.dot(np.transpose(z),z));print()
print(np.dot(X,Y));print()
```

The naming of the `dot` command is unfortunate because it doesn't actually perform a dot product! It performs a matrix/matrix multiplication, where vectors are treated as 1D matrices. If you run np.dot(z,z) you will get an error, even then $\langle z,z\rangle$ is a valid linear algebra operation.

Fortunately, you can avoid `dot` altogether. A much cleaner approach is to use numpy's `@` operator. This operator works directly on 2D ndarrays with compatible dimensions.

```python
print(X@z);print()
print(np.transpose(z)@z);print()
print(X@Y)
```

Note that `@` between a row and column vector computes an outerproduct instead of an inner product.

```python
print(z@np.transpose(z))
```

The @ command can be used with vectors that are 1-d not 2-d arrays. For example, the following code is valid. But note that the output is a 1-d, not a 2-d array

```python
X = np.array([[1,2],[3,4],[5,6]])
z = np.array([2,3])
w = np.array([2,3,4])
print(X@z)
print()
print(w@X)
```
